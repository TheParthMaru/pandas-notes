### Working on Dataframes

```python
# View the entire dataframe

print(coffee_data)
```

```python
# Better way

display(coffee_data)
```

```python
# Access random data

coffee_data.sample(3) # By defeault it returns single random row.
```

```python
# loc and iloc

'''
- They are two main ways to select rows and columns from a dataframe.
- loc is label based. We use index names and column names.
- iloc is position based. We use integer positions (0,1,2,3,4,...) like list indexing.
- The biggest difference is slicing.
'''

# Syntax
# coffee_data.loc[rows, columns]

coffee_data.loc[4] # Displays the data of the row index 4.

# For better view, coffee_data.loc[[4]]
```

```python
coffee_data.loc[[0,2,3]] # Display the data of index 0, 2 and 3
```

```python
coffee_data.loc[0:5] # Added slicing
```

```python
# Grabbing the column along with the rows

coffee_data.loc[0:5, ["Day", "Units Sold"]]
```

```python
display(coffee_data)
```

```python
# Using iloc

coffee_data.iloc[0]
```

```python
coffee_data.iloc[0:5, 2]
```

```python
# Adding a square brackets makes a difference in the view
coffee_data.iloc[0:5, [2]]
```

```python
coffee_data.iloc[0:5, [0,2]]
```

```python
# One of the difference to observe is that with loc the end index is inclusive. But with iloc, it is exclusive.
# iloc only works with indices for both rows and columns.
```

```python
# Updating a specific value using loc

coffee_data.loc[1, "Units Sold"] = 69

display(coffee_data)

# We can update multiple values by slicing.
```

```python
# We can also get specific values using at and iat
# Slicing won't work. Only single values.

print(coffee_data.at[0, "Units Sold"])
```

```python
coffee_data.iat[0,0]
```

```python
# Another way to grab columns is just by using their names

coffee_data.Day
```

```python
coffee_data["Units Sold"]
```

```python
# Sorting the values

coffee_data.sort_values("Units Sold", ascending=True) # Sorting in ascending order
```

```python
coffee_data.sort_values("Units Sold", ascending=False) # Sorting in descending order
```

```python
coffee_data.sort_index(ascending=False) # Sorting the indices in descending order
```

```python
# Multi level sorting and filtering
coffee_data.sort_values(["Units Sold", "Coffee Type"], ascending=False)

# First sort by units sold. If there are two coffee types with the same value, then it will sort those as well.
```

```python
coffee_data.sort_values(["Units Sold", "Coffee Type"], ascending=[0,1])

# This means sort the units sold in descending order and coffee type in ascending order
```

## Filtering Data

```python
# Loading the olympics data

bios = pd.read_csv("data/bios.csv")
noc_regions = pd.read_csv("data/noc_regions.csv")
results = pd.read_csv("data/results.csv")
```

```python
bios.head()
```

```python
bios.info()
```

```python
# Find atheletes height greater than 200 cm

bios.loc[bios["height_cm"] > 200.00, ["name", "height_cm"]]
```

```python
# We can even do the above thing without loc

bios[bios["height_cm"] > 200.00][["name", "height_cm"]]
```

```python
# Adding more conditions
# Find the atheletes having height > 200 cm and born in UK.


bios.loc[(bios["height_cm"] > 200.00) & (bios["born_country"] == "GBR")]
```

```python
# Find atheletes whose name contains "Keith"

bios[bios["name"].str.contains("Keith")][["name", "born_country"]]

# Inside the contains, you can add a flag case=False, and then just search with "keith". It will give all possible "keith" variants.
# If you want to search for two names like keith or patrick, use .contains("keith|patrick"). It is a regex thing.
```

```python
# Filtering using isin
# Find atheles born in USA, France and Great Britain.

bios[bios["born_country"].isin(["USA", "FRA", "GBR"])][["name", "born_country"]]
```

```python
# In the couple of cells above, we can see that we are repeating bios again and again in the same statement to filter our meaningful data. We can reduce the repetition by using query.

bios.query("born_country == 'USA' and born_city == 'Seattle'")

# I don't see much options to filter the data using query. Our traditional way allowed us to drill down in a better way.
```
